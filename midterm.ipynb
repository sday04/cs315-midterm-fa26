```python
# Define the function "calculate_days"
import re
import datetime
import dateutil
from dateutil import parser

def calculate_days(text):
    # pattern matching
    pattern = re.compile(r"\w{3}\s\d{2},\s\d{4}")
    matches = pattern.findall(text)

    # start and end dates

    startDate = parser.parse(matches[0])
    endDate = parser.parse(matches[1])

    numDays = (endDate-startDate).days
    return numDays
    
```

```python
phrase = 'Aired: Jun 18, 2022 - Aug 28, 2022'
# test that your function works

results = calculate_days(phrase)

print(results)
```
